```python
import pandas as pd
import numpy as np




df_vendas = pd.read_csv('vendas_tech.csv', sep=',', encoding="utf8", low_memory=False)
df_lojas = pd.read_excel("gerentes_lojas.xlsx")

display(df_vendas)
display(df_lojas)
```

```python
#inspeção
display(df_vendas.info())
display(df_vendas.describe())





```

```python
#tratamento de dados




#colunas

df_analise = df_vendas.drop(columns = ["Data_Base"])



#nulos
print((df_analise["Loja"].isnull().sum() * 100)/ len(df_analise["Loja"]), "% de nulos na coluna Loja")
"""Verificar a origem desse valor nulo, se for um erro de digitação ou se realmente não há loja associada a esse pedido. Caso seja um erro de digitação, podemos preencher com o valor correto.
 Caso não haja loja associada, podemos preencher com "Online" ou "Sem Loja". """
print(df_analise.isnull().sum(), "\n")
df_analise["Loja"] = df_analise["Loja"].fillna("Online")

#dtype errados
df_analise["Data_Formatada"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d")
#

#padronização
colunas_p_padronizar = ["Loja", "Produto"]
for col in colunas_p_padronizar:
    df_analise[col] = df_analise[col].str.strip().str.title()
display(df_analise.head())




#duplicatas

df_analise = df_analise.drop_duplicates(subset=["ID_Pedido"])
df_analise.info()
```

```python
# criar novas colunas

#faturamento
df_analise["Faturamento"] = df_analise["Preco_Unitario"] * df_analise["Qtd"]
df_analise

#regiao
#print(df_analise["Loja"].value_counts())
print(df_analise["Loja"].unique())

dict_reg ={'São Paulo':'sudeste',
         'Belo Horizonte':'sudeste',
         'Online':'Online',
         'Rio De Janeiro':'sudeste',
         'Salvador':'nordeste',
         'Recife':'nordeste',
         'Curitiba':'sul',
        'Porto Alegre':'sul'}

df_analise["Regiao"] = df_analise["Loja"].map(dict_reg)

# forma de venda

df_analise["Forma_de_venda"] = np.where(df_analise["Loja"] == "Online", "Online", "Presencial")
df_analise["Regiao"] = df_analise["Regiao"].str.title()

display(df_analise)
```

```python
#Analisar

#ordernar colunas
df_analise = df_analise.sort_values(by="Data_Formatada")
df_analise = df_analise.reset_index(drop = True) #-> reseta o index da base de dados, e o drop remove a coluna de indices antiga

#-Exportar pedaços da base

#Localizar uma informação em específica
#.loc, .iloc, condicional

#.loc = localiar algo de acordo com a coluna e o indice tipo select coluna where..
id_pedido = 67
loja = df_analise.loc[id_pedido - 1, "Loja"]
produto = df_analise.loc[id_pedido -1, "Produto"]
cliente = df_analise.loc[id_pedido -1, "Cliente"]
print("Loja =", loja)
print("Produto =",produto)
print("Cliente =", cliente)

#.iloc -> por posição
loja = df_analise.iloc[66,2 ]
produto = df_analise.iloc[66,3 ]
cliente = df_analise.iloc[66, 6]
print("Loja =", loja)
print("Produto =",produto)
print("Cliente =", cliente)

#condiucional tipo where 
df_Vendas_SP = df_analise[df_analise["Loja"]== "São Paulo"]
df_Vendas_SP = df_Vendas_SP[["Loja", "Faturamento"]]
df_Vendas_SP["Faturamento_Total"] = df_Vendas_SP["Faturamento"].sum()
display(df_Vendas_SP)
df_Vendas_SP.to_csv("df_Vendas_SP.csv")
```
